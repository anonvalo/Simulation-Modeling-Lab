Student's Assignment 1: Basic Matrix Operations
1. Create two 3×3 matrices with random integers in Python.
2. Perform addition, subtraction, and multiplication operations on these matrices.
3. Compute the determinant, inverse, and rank of one of the matrices.

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
import random
```

```python
#create matrix
R  = np.random.rand()
A = np.random.randint(5,30,(3,3))
print("Matrix_A\n", A)
B = np.random.randint(4,16,(3,3))
print("Matrix_B\n", B)

#Summation
mat_sum = A+B
print("Matrix Summation\n", mat_sum)

#Subtraction
mat_sub = A-B
print("Matrix Subtraction\n", mat_sub)

#Multiplication
mat_prod = A * B
print("Matrix Product\n", mat_prod)


#detarminant
det_A = np.linalg.det(A)
print("Detarminant of A\n",det_A)


#rank
rank_A = np.linalg.matrix_rank(A)
print("Rank of A \n", rank_A)


#inverse
if np.linalg.det(A)!= 0:
  inv_A = np.linalg.inv(A)
else:
  inv_A = "Not a matrix"

print("Inverser Matrix\n", inv_A)
```
