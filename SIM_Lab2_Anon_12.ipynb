**Analyze Matrix A**

*   Shape of A
*   Determinant
*   Rank
*   Eigenvalues

*   Inverse (only if determinant ≠ 0)

**ID**:**40**


```python
d1=2 #last digit of my id
d2=1 #second last digit of my id
A = np.array([[d1+2,d2+1], [2*d1, d2+2]])

#Shape of A
shape_A=A.shape
#Determinant of matrix A
det_A = np.linalg.det(A)
#Rank of matrix A
rank_A = np.linalg.matrix_rank(A)
#Eigenvalues of matrix A
eigenvalue_A = np.linalg.eigvals(A)
#Inverse of matrix A
if det_A!=0:
  invs_A = np.linalg.inv(A)
else:
  invs_A = "The matrix is singular, Cannot compute"

print("Shape of  matrix A: \n",shape_A)
print("Determinant of  matrix A: \n",det_A)
print("Rank of  matrix A: \n",rank_A)
print("Eigenvalue of  matrix A: \n",eigenvalue_A)
print("Inverse of  matrix A: \n",invs_A)
```

**Create a new matrix B by changing exactly one value in A**:


*   Shape of A
*   Determinant
*   Rank
*   Eigenvalues

*   Inverse (only if determinant ≠ 0)



```python
d1=0 #last digit of my id
d2=4 #second last digit of my id
A = np.array([[d1+2,d2+1], [2*d1, d2+2]])

B = np.array([[d1+2,d2+1], [2*d1+1, d2+2]]) #Changing first element fo second raw from 2*d1

#Shape of B
shape_B=B.shape

#Determinant of matrix B
det_B = np.linalg.det(B)

#Rank of matrix B
rank_B = np.linalg.matrix_rank(B)

#Eigenvalues of matrix B
eigenvalue_B = np.linalg.eigvals(B)

#Inverse of matrix B
if det_B!=0:
  invs_B = np.linalg.inv(B)
else:
  invs_B = "The matrix is singular, Cannot compute"

print(B)
print("Shape of  matrix A: \n",shape_B)
print("Determinant of  matrix A: \n",det_B)
print("Rank of  matrix A: \n",rank_B)
print("Eigenvalue of  matrix A: \n",eigenvalue_B)
print("Inverse of  matrix A: \n",invs_B)
```

**1. How did the determinant change and why?**

Ans : The determinant of matrix A is 12. The determinant of matrix B is 7. It changed because the determinant formula is [(A11 * A22)- (A12 * A21)]. So in Matrix A, the value of A21 is 0 and A is a 2x2 matrix. SO the determinant of A is based on (A11*A22). Matrix B = A. But in matrix B the value of B21 changed to 1. So the determinant of B is based on [(B11 * B22)- (B12 * B21)].



**2. Did the rank change?**

**Ans :** The rank of matrix A is 2. And the rank of matrix B is also 2. rank means the maximum number of linearly independent rows or columns in the matrix.
In both matrix the determinant is non-zero. Non-zero determinant means rows and columns are linearly independent.
Therefore, both matrices are full rank. So even though values changed, neither matrix lost independent information.

**3. How did the eigenvalues respond to the value change?**
**Ans : The eigenvalues changed after we change the matrix value.
